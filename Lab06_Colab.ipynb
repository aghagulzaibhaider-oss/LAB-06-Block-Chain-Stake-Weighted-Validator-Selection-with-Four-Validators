# Lab 06 - Stake-Weighted Validator Selection (4 validators)
Run the cells from top to bottom. Each task cell is a separate run.

**Purpose:** show that in Proof of Stake, a validator's chance of being selected is proportional to its stake, and that slashing reduces that chance. (Toy simulation, not a real PoS protocol.)

## Step 0 - Setup (run first)

```
import collections, random

def run_lab(rounds=2000, slash=False, seed=2026):
    assert 1 <= rounds <= 100000, "rounds must be between 1 and 100000"
    stake = {"Amina": 50, "Bilal": 30, "Chen": 20, "Dalia": 10}   # Dalia = extension
    if slash:
        print("DEMO: validator Amina violated toy rules: 20 units slashed")
        stake["Amina"] -= 20
    rng = random.Random(seed)
    choices = rng.choices(list(stake), weights=list(stake.values()), k=rounds)
    counts = collections.Counter(choices)
    total = sum(stake.values())
    for name, amount in stake.items():
        print(f"{name:6} stake={amount:3} expected={amount/total:6.1%} observed={counts[name]/rounds:6.1%}")
    assert sum(counts.values()) == rounds
    print("EXPECTED SHARE SUM:", total, "stake units")
    print("PASS: stake-weighted selection and optional penalty modeled.")
    print("CAUTION: validator selection in real chains depends on protocol and many additional checks.")
```

## Task 5 - Optional: bar chart of expected vs observed

```
import matplotlib.pyplot as plt
stake = {"Amina": 50, "Bilal": 30, "Chen": 20, "Dalia": 10}
rounds = 10000
rng = random.Random(2026)
counts = collections.Counter(rng.choices(list(stake), weights=list(stake.values()), k=rounds))
total = sum(stake.values())
names = list(stake)
exp = [stake[n]/total*100 for n in names]
obs = [counts[n]/rounds*100 for n in names]
x = range(len(names))
plt.bar([i-0.2 for i in x], exp, 0.4, label="Expected %")
plt.bar([i+0.2 for i in x], obs, 0.4, label="Observed %")
plt.xticks(list(x), names); plt.ylabel("Selection share (%)")
plt.title("Stake-weighted selection, 10,000 draws"); plt.legend(); plt.show()
```
